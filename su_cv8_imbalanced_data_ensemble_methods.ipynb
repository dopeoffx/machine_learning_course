# Imbalanced dataset

```python
df = pd.read_csv('datasets/nonsep.csv', sep=';', header=None)
df.columns = ['x', 'y', 'class']
df_imbalanced = df[df['class'] == -1].copy()
# df_imbalanced = df_imbalanced.append(df[df['class'] == 1].sample(n=10, random_state=42))
df_imbalanced = pd.concat([df_imbalanced, df[df['class'] == 1].sample(n=10, random_state=42)], ignore_index=True)
print(df.shape)
print(df_imbalanced.shape)
```

```python
sns.scatterplot(data=df_imbalanced, x='x', y='y', hue='class')
plt.title('Imbalanced dataset')
```

## Train test split for evaluation
Be aware to do under/oversampling only for training split. Don't disturb class ratio in you test set of data. U want to have test set isolated from under/oversampling procedures, to mimic real data distribution for classificator evaluation. You add a lot of bias to the evaluation, when using under/oversampling methods for whole dataset.

```python
df_train, df_test = train_test_split(df_imbalanced, test_size=0.5, stratify=df_imbalanced['class'], random_state=42)
X_train, X_test, y_train, y_test = df_train[df_train.columns[df_train.columns != 'class']].values, df_test[df_test.columns[df_test.columns != 'class']].values, df_train['class'].values, df_test['class'].values
```

```python
sns.scatterplot(x=X_train[:, 0], y=X_train[:, 1], hue=y_train)
plt.title('Train set')
```

#### Class distribution in train set:

```python
pd.Series(y_train).value_counts()
```

```python
clf = LinearSVC()
clf.fit(X_train, y_train)
print(f'train score is {f1_score(y_train, clf.predict(X_train))}')
print(f'test score is {f1_score(y_test, clf.predict(X_test))}')
```

```python
sns.scatterplot(x=X_test[:, 0], y=X_test[:, 1], hue=y_test, style=clf.predict(X_test), markers=['o','s'])
plt.title('Evaluation of predictions on test set')
```

### Bagging (Bootstrap Aggregating)

Bagging je ensemble metoda, která zvyšuje stabilitu a přesnost prediktivních modelů tím, že trénuje více modelů (tzv. base learners) na náhodně vybraných podmnožinách trénovacích dat vytvořených metodou bootstrap (vzorkování s opakováním).
Každý model se učí na jiné kombinaci dat a jejich výsledky jsou následně zprůměrovány (u regrese) nebo se provede hlasování (u klasifikace).
Tím se snižuje variabilita modelu a omezuje riziko přeučení.
Typickým příkladem baggingu je algoritmus Random Forest, který kombinuje mnoho rozhodovacích stromů.

### Boosting

Boosting je ensemble metoda, která kombinuje více „slabých“ modelů (weak learners) do jednoho silného (strong learner).
Modely jsou trénovány sekvenčně, přičemž každý další model se snaží opravovat chyby, které udělaly modely předchozí.
Jednotlivé modely mají váhy podle své úspěšnosti, což umožňuje systému více se soustředit na obtížně predikovatelné případy.
Boosting obvykle snižuje chybu (bias) modelu a vede ke zlepšení přesnosti.
Typickými zástupci jsou AdaBoost, Gradient Boosting, XGBoost nebo LightGBM.

### Stacking (Stacked Generalization)

Stacking je ensemble technika, která kombinuje výstupy několika různých modelů (např. logistická regrese, rozhodovací strom, SVM) pomocí tzv. meta-modelu.
Tento meta-model se učí na výstupech (predikcích) základních modelů a snaží se nalézt optimální způsob, jak jejich výsledky zkombinovat.
Cílem je využít silné stránky jednotlivých modelů a tím dosáhnout vyšší celkové přesnosti.
Na rozdíl od baggingu a boostingů zde modely nemusí být stejného typu – stacking kombinuje různé přístupy.

```python
rf = RandomForestClassifier(
    n_estimators=300,        
    max_depth=None,        
    n_jobs=-1, random_state=42
)
rf.fit(X_train, y_train)
print(f'train score is {f1_score(y_train, rf.predict(X_train))}')
print(f'test score is {f1_score(y_test, rf.predict(X_test))}')
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    n_estimators=300,     # kolik slabých učících (stromů)
    learning_rate=0.05,  # menší krok → víc stromů, stabilnější učení
    max_depth=3,         # stromky držíme malé
    random_state=42
)
gb.fit(X_train, y_train)
print(f'train score is {f1_score(y_train, gb.predict(X_train))}')
print(f'test score is {f1_score(y_test, gb.predict(X_test))}')
```

```python
from sklearn.svm import SVC

base_learners = [
    ("rf", RandomForestClassifier(n_estimators=300, random_state=42)),
    ("svc", SVC(kernel="rbf", C=2, probability=True, random_state=42))
]
meta = LogisticRegression(max_iter=1000, random_state=42)

stack = StackingClassifier(
    estimators=base_learners,
    final_estimator=meta,
    stack_method="predict_proba",  # meta-model dostane pravděpodobnosti
    n_jobs=-1
)
stack.fit(X_train, y_train)
print(f'train score is {f1_score(y_train, stack.predict(X_train))}')
print(f'test score is {f1_score(y_test, stack.predict(X_test))}')
```
